# 리스트(List) 예시

```python
fruits = ["사과", "바나나", "포도", "딸기"]

print("전체 리스트:", fruits)
print("첫 번째 과일:", fruits[0])
```

```python
fruits.append("수박")

print("수박 추가 후:", fruits)
```

# 튜플(Tuple) 예시

```python
colors = ("빨강", "파랑", "노랑", "초록")

print("전체 튜플:", colors)
print("첫 번째 색상:", colors[0])
print("두 번째 색상:", colors[1])
```

# 셋(Set) 예시

```python
numbers = {1, 2, 3, 3, 4, 4, 5}

print("전체 셋:", numbers)
```

```python
numbers.add(6)
print("6 추가 후:", numbers)
```

# 딕셔너리(Dictionary) 예시

```python
student = {
    "이름": "이재현",
    "나이": 42,
    "학과": "인공지능학과",
    "학년": 4
    }

print("전체 딕셔너리:", student)
print("이름:", student["이름"])
print("학과:", student["학과"])
```

```python
student["학교"] = "세종사이버대학교"

print("학교 추가 후:", student)
```
